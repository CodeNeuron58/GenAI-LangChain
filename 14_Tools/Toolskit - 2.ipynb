```python
from langchain_core.tools import tool
```

```python
# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
# MathToolkit class definition with named access
class MathToolkit:
    def __init__(self):
        self.tools = {
            "add": add,
            "multiply": multiply
        }
        
    def get_tool(self, tool_name: str):
        return self.tools.get(tool_name)
```

```python
# Instantiate MathToolkit
math_toolkit = MathToolkit()

# Example usage
a = 5
b = 3
```

```python
# Access tools by name and perform operations
add_tool = math_toolkit.get_tool("add")
multiply_tool = math_toolkit.get_tool("multiply")
```

```python
add_tool
```

```python
add_result = add_tool.invoke({"a": a, "b": b})
multiply_result = multiply_tool.invoke({"a": a, "b": b})
```

```python
print("Addition:", add_result)
```

```python
print("Multiplication:", multiply_result)
```
